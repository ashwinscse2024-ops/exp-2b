```python
import pandas as pd

data = {
    'name': ['Ashwin', 'Rahul', 'Priya', 'Kiran', 'Anu'],
    'age': [20, 21, 20, 22, 21],
    'marks': [85, 72, 90, 65, 78],
    'category': ['A', 'B', 'A', 'B', 'A']
}

df = pd.DataFrame(data)

df.to_csv('data.csv', index=False)

print("data.csv created successfully")
```

```python
df = pd.read_csv('data.csv')
print("First 5 rows:\n", df.head())
print("\nLast 5 rows:\n", df.tail())
```

```python
print("\nData Types:\n", df.dtypes)
df.info()
```

```python
print("\nSummary Statistics:\n", df.describe())
```

```python
df.fillna(df.mean(numeric_only=True), inplace=True)

df['new_marks'] = df['marks'] * 2
series = df['marks']
print("\nSeries Addition:\n", series + 10)
```

```python
filtered_df = df[
    (df['marks'] > 50) &
    (df['age'] < 22)
]

print("\nFiltered DataFrame:\n", filtered_df)
```

```python
grouped = df.groupby('category')['marks'].mean()

print("\nGrouped Mean:\n", grouped)
```

```python
df_sorted = df.sort_values(
    by='marks',
    ascending=False
)

print("\nSorted DataFrame:\n", df_sorted)
```

```python
masked_df = df[
    df['marks'] > df['marks'].median()
]

print("\nMasked DataFrame:\n", masked_df)
```

```python
df.drop_duplicates(inplace=True)
df.dropna(inplace=True)

subset_df = df[['name', 'marks']]

print("\nSubset DataFrame:\n", subset_df)

subset_df.to_csv('filtered_data.csv', index=False)
```
